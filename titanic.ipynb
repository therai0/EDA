## EDA on Titanic Dataset 

```python
import pandas as pd 
import seaborn as sns 
import numpy as np 
import matplotlib.pyplot as plt 
```

### Loading dataset and basic overview

```python
df = sns.load_dataset("titanic")
```

##  Univariant analysis

### Survived columns 

+ Number of servival is less comperatively 

```python
df["survived"].value_counts()
```

```python
sns.countplot(x="survived", data=df)
plt.xlabel("0:died 1:survived")
plt.ylabel("Counts")
plt.show()
```

```python
plt.title("Pie Chart of Survival column")
plt.pie([549,342],autopct="%1.1f%%")
plt.show()
```

### Plcass
+ Most of the people are belong to class 3

```python
df["pclass"].value_counts()
```

```python
sns.countplot(x="pclass",data=df)
plt.xlabel("Classes")
plt.ylabel("counts")
plt.show()
```

### Sex

+ The number of male is more then female 

```python
df["sex"].value_counts()
```

```python
sns.countplot(x="sex",data=df)
plt.xlabel("Sex")
plt.ylabel("counts")
```
